```python

import random
import numpy as np
import math

def fitness(x):
    return x**2 + 6*x - 2


class Cuckoo:
    def __init__(self, position):
        self.position = position
        self.fitness = fitness(position)

    def generate_new_solution(self, alpha, beta):
        sigma = (
            math.gamma(1 + beta) * math.sin(math.pi * beta / 2)
            / (
                math.gamma((1 + beta) / 2)
                * beta
                * 2**((beta - 1) / 2)
            )
        ) ** (1 / beta)

        u = np.random.normal(0, sigma)
        v = np.random.normal(0, 1)

        levy = u / (abs(v)**(1 / beta))

        new_position = self.position + alpha * levy
        new_position = max(-10, min(10, new_position))

        return Cuckoo(new_position)


class CuckooSearch:
    def __init__(self, num_nests, max_iterations, pa, beta):
        self.num_nests = num_nests
        self.max_iterations = max_iterations
        self.pa = pa
        self.beta = beta
        self.nests = []

        for i in range(num_nests):
            position = random.uniform(-10, 10)
            self.nests.append(Cuckoo(position))

    def optimize(self):
        best_nest = min(self.nests, key=lambda x: x.fitness)

        alpha = 0.1

        for iteration in range(1, self.max_iterations + 1):

            for i in range(self.num_nests):

                new_nest = self.nests[i].generate_new_solution(
                    alpha, self.beta
                )

                j = random.randint(0, self.num_nests - 1)

                if new_nest.fitness < self.nests[j].fitness:
                    self.nests[j] = new_nest

            self.nests.sort(key=lambda x: x.fitness)

            num_abandoned = int(self.pa * self.num_nests)

            for i in range(num_abandoned):
                position = random.uniform(-10, 10)
                self.nests[-(i + 1)] = Cuckoo(position)

            self.nests.sort(key=lambda x: x.fitness)

            if self.nests[0].fitness < best_nest.fitness:
                best_nest = self.nests[0]

            if iteration == 1 or iteration % 10 == 0:
                print("Iteration:", iteration)
                print("Best x:", best_nest.position)
                print("Best fitness:", best_nest.fitness)
                print()

        print("FINAL RESULT")
        print("Best x:", best_nest.position)
        print("Minimum fitness:", best_nest.fitness)


if __name__ == '__main__':

    random.seed(42)
    np.random.seed(42)

    grid_size = 10
    num_nests = grid_size * grid_size
    max_iterations = 100
    pa = 0.25
    beta = 1.5

    cs = CuckooSearch(
        num_nests,
        max_iterations,
        pa,
        beta
    )

    cs.optimize()
```
