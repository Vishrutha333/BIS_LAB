```python
import random

def fitness(x):
    return (x[0] - 5)**2 + (x[1] - 2)**2

def gwo():
    population = 12
    max_iter = 100
    dim = 2
    lower = 0
    upper = 10

    wolves = [
        [random.uniform(lower, upper) for _ in range(dim)]
        for _ in range(population)
    ]

    for iteration in range(max_iter):
        wolves.sort(key=fitness)

        alpha = wolves[0][:]
        beta = wolves[1][:]
        delta = wolves[2][:]

        a = 2 - 2 * iteration / max_iter
        new_wolves = []

        for wolf in wolves:
            new_position = []

            for j in range(dim):
                r1, r2 = random.random(), random.random()
                A1 = 2 * a * r1 - a
                C1 = 2 * r2
                D_alpha = abs(C1 * alpha[j] - wolf[j])
                X1 = alpha[j] - A1 * D_alpha

                r1, r2 = random.random(), random.random()
                A2 = 2 * a * r1 - a
                C2 = 2 * r2
                D_beta = abs(C2 * beta[j] - wolf[j])
                X2 = beta[j] - A2 * D_beta

                r1, r2 = random.random(), random.random()
                A3 = 2 * a * r1 - a
                C3 = 2 * r2
                D_delta = abs(C3 * delta[j] - wolf[j])
                X3 = delta[j] - A3 * D_delta

                position = (X1 + X2 + X3) / 3
                position = max(lower, min(upper, position))
                new_position.append(position)

            new_wolves.append(new_position)

        wolves = new_wolves

    wolves.sort(key=fitness)
    best = wolves[0]

    print("Best solution:")
    print("x1 =", round(best[0], 6))
    print("x2 =", round(best[1], 6))
    print("Minimum function value =", round(fitness(best), 8))

gwo()
```
