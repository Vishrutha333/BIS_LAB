```python

import random

def fitness(x):
    return x**2 + 6*x - 2


def initialize_population(grid_size, lower_bound, upper_bound):
    grid = []
    for i in range(grid_size):
        row = []
        for j in range(grid_size):
            row.append(random.uniform(lower_bound, upper_bound))
        grid.append(row)
    return grid


def get_neighbors(grid, i, j):
    neighbors = []
    for x in range(max(0, i-1), min(len(grid), i+2)):
        for y in range(max(0, j-1), min(len(grid[0]), j+2)):
            if x != i or y != j:
                neighbors.append(grid[x][y])
    return neighbors


def update_states(grid, lower_bound, upper_bound):
    new_grid = []

    for i in range(len(grid)):
        row = []
        for j in range(len(grid[0])):
            neighbors = get_neighbors(grid, i, j)

            neighbor_mean = sum(neighbors) / len(neighbors)
            current_fitness = fitness(grid[i][j])
            neighbor_fitness = sum(fitness(x) for x in neighbors) / len(neighbors)

            if neighbor_fitness < current_fitness:
                new_value = neighbor_mean
            else:
                new_value = grid[i][j]

            new_value = max(lower_bound, min(upper_bound, new_value))
            row.append(new_value)

        new_grid.append(row)

    return new_grid


def parallel_cellular_algorithm():
    grid_size = 10
    lower_bound = -10
    upper_bound = 10
    iterations = 100

    grid = initialize_population(grid_size, lower_bound, upper_bound)

    best_x = None
    best_fitness = float('inf')

    for iteration in range(1, iterations + 1):

        for i in range(grid_size):
            for j in range(grid_size):
                current_fitness = fitness(grid[i][j])

                if current_fitness < best_fitness:
                    best_fitness = current_fitness
                    best_x = grid[i][j]

        if iteration == 1 or iteration == 100:
            print("Iteration", iteration)
            print("Best x =", best_x)
            print("Best fitness =", best_fitness)
            print()

        grid = update_states(grid, lower_bound, upper_bound)

    for row in grid:
        for x in row:
            if fitness(x) < best_fitness:
                best_fitness = fitness(x)
                best_x = x

    print("Final Result:")
    print("Best x =", best_x)
    print("Best fitness =", best_fitness)


if __name__ == '__main__':
    parallel_cellular_algorithm()
```
